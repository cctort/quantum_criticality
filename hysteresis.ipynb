```python
def Z_from_S(S_iw_dict, inputs_dict):

    Z_dict = {}
    for key, S in S_iw_dict.items():
        Z_dict[key], _, _ = get_Z(S, 1/dict(key)['T'], inputs_dict['n_iw'])
    
    return Z_dict

(S_iw, diff_m), inputs = HDFread(f'data/{square.dim}d/hysteresis/metal_n05.h5', ['S_iw', 'diff'])
Zm_dict = Z_from_S(S_iw, inputs)


(S_iw, diff_a), inputs = HDFread(f'data/{square.dim}d/hysteresis/atom_n05.h5', ['S_iw', 'diff'])
Za_dict = Z_from_S(S_iw, inputs)

common_keys = sorted(set(Zm_dict.keys()) & set(Za_dict.keys()))

common_params, Zm_list, Za_list, diffm_list, diffa_list = [], [], [], [], []
for key in common_keys:
    common_params.append((dict(key)['U'], dict(key)['T'], dict(key)['n']))
    Zm_list.append(Zm_dict[key])
    Za_list.append(Za_dict[key])
    diffm_list.append(diff_m[key])
    diffa_list.append(diff_a[key])

title = f'Square lattice (n = {n})'
plot1 = plot_diagram(common_params, np.abs(np.array(Zm_list) - np.array(Za_list)), z_label=r'$Z_{metal}-Z_{atom}$', point_label=('U','T'), title=title, shape=(1,2,1))
plot2 = plot_diagram(common_params, diffm_list, z_label=r'$\Sigma$ residuals (from $U=0$)', point_label=('U','T'), title=title, logscale=True, shape=(2,2,2), plot=plot1)
plot_diagram(common_params, diffa_list, z_label=r'$\Sigma$ residuals (from $U=\infty$)', point_label=('U','T'), title=title, logscale=True, shape=(2,2,4), plot=plot2)


plt.savefig(f'data/{square.dim}d/hysteresis/pd_n05.png', dpi=100, bbox_inches='tight')
plt.show()
```
